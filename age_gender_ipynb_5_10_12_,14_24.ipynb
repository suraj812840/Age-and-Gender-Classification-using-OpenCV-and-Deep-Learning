```python
fd=FER()
```

```python
px=plt.imread("/content/20_1_4_20170103223121735.jpg")
```

```python
px.shape
```

```python
px.size
```

```python
plt.imshow(px)
plt.show()
```

```python
x,y,w,h=fd.find_faces(px)[0]
```

```python
x,y,w,h
```

```python
# Create a copy of the image
px_copy = px.copy()

# Draw the rectangle on the copy
cv.rectangle(px_copy, (x, y), (x + w, y + h), [255, 0, 0], 1)

# Display the modified copy
plt.imshow(px_copy)
plt.show()
```

```python
px_face=px[y:y+h,x:x+w]
plt.imshow(px_face)
plt.show()
```

```python
px_gray=cv.cvtColor(px_face,cv.COLOR_BGR2GRAY)
plt.imshow(px_gray,cmap="gray")
plt.show()
```

```python
plt.figure(figsize=[2,2])
px_final=cv.resize(px_gray,(48,48))
plt.imshow(px_final,cmap="gray")
plt.show()
```

```python
from keras.models import Sequential
from keras.layers import Dense,Activation,Flatten
```

```python
model=Sequential()
model.add(Flatten(input_shape=[48,48]))
model.add(Dense(units=100,use_bias=False))
model.add(Activation('relu'))
model.add(Dense(200,use_bias=False))
model.add(Activation('relu'))
model.add(Dense(2,use_bias=False))
model.add(Activation('sigmoid'))
```

```python
model.summary()
```

```python
model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
```
